# The number of objects in each class

```python
from collections import Counter

CLASS_NAMES = {
    0: "person",
    1: "umbrella",
    2: "bottle",
    3: "cup",
    4: "fork",
    5: "spoon",
    6: "chair",
    7: "cell phone",
    8: "book",
}

def count_classes(label_dir):

    counts = Counter()

    for label_file in label_dir.glob("*.txt"):

        with open(label_file, "r") as f:

            for line in f:

                parts = line.strip().split()

                if len(parts) == 5:
                    class_id = int(parts[0])
                    counts[class_id] += 1

    return counts


train_counts = count_classes(
    DATASET_DIR / "labels" / "train"
)

val_counts = count_classes(
    DATASET_DIR / "labels" / "val"
)

print("TRAIN")
print("-" * 40)

for class_id, name in CLASS_NAMES.items():
    print(
        f"{class_id}: {name:12s} "
        f"{train_counts[class_id]:,}"
    )

print("\nVALIDATION")
print("-" * 40)

for class_id, name in CLASS_NAMES.items():
    print(
        f"{class_id}: {name:12s} "
        f"{val_counts[class_id]:,}"
    )
```

## I'll try printing the object sizes by class

```python
from collections import defaultdict
import numpy as np

class_names = {
    0: "person",
    1: "umbrella",
    2: "bottle",
    3: "cup",
    4: "fork",
    5: "spoon",
    6: "chair",
    7: "cell phone",
    8: "book",
}

# Store normalized bounding-box areas
areas = defaultdict(list)

label_dir = DATASET_DIR / "labels" / "train"

for label_file in label_dir.glob("*.txt"):

    with open(label_file, "r") as f:

        for line in f:

            parts = line.strip().split()

            if len(parts) != 5:
                continue

            class_id = int(parts[0])

            if class_id not in class_names:
                continue

            width = float(parts[3])
            height = float(parts[4])

            area = width * height

            areas[class_id].append(area)


print("Bounding box statistics")
print("-" * 70)

for class_id, name in class_names.items():

    values = np.array(areas[class_id])

    print(
        f"{name:12s} "
        f"count={len(values):6,} "
        f"mean={values.mean():.5f} "
        f"median={np.median(values):.5f} "
        f"min={values.min():.5f}"
    )
```
